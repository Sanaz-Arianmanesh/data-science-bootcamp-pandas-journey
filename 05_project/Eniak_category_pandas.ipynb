# Searching for the category

```python
import pandas as pd

url = "https://drive.google.com/uc?export=download&id=12SsiaeVBNraWybslbDJRn_9dSgLU0W4d"
products_qu = pd.read_csv(url)
```

```python
products_cat = products_qu.copy()
```

### Comment: Remove of the promo_price column

```python
# The promo_price column has messy, non-numeric values from the CSV.
# Because the real selling prices are already in orderlines.unit_price, this column is not reliable for analysis and was removed.

products_cat = products_cat.drop(columns=["promo_price"])
products_cat.head()
```

```python
products_cat.head()
```

```python
#  Creating a column category with a blank string "".

products_cat["category"] = ""
products_cat.head()
```

```python
# Refurbished flag (not a category)
products_cat["is_refurbished"] = products_cat["desc"].str.contains(
    "refurbished|reconditioned|like new",
    case=False,
    na=False
)

# Create simple business categories
products_cat["category"] = "other"

# Devices (high value products)
products_cat.loc[
    products_cat["name"].str.contains("iphone|ipad|macbook|imac|mac mini|mac pro", case=False, na=False),
    "category"
] = "device"

# Peripherals
products_cat.loc[
    products_cat["desc"].str.contains("keyboard|mouse|trackpad|headset|headphone|monitor|speaker", case=False, na=False),
    "category"
] = "peripheral"

# Components
products_cat.loc[
    products_cat["desc"].str.contains("ram|ssd|hard drive|hard disk|storage|memory", case=False, na=False),
    "category"
] = "component"

# Accessories
products_cat.loc[
    products_cat["desc"].str.contains("cable|adapter|case|dock|charger|backpack|stand|mStand", case=False, na=False),
    "category"
] = "accessory"

# Network / server
products_cat.loc[
    products_cat["desc"].str.contains("nas|server|synology|raid", case=False, na=False),
    "category"
] = "network/server"

products_cat["category"].value_counts()
```

```python
# # Display the first 10 products showing SKU, product name,
# the assigned category, and whether the product is refurbished
products_cat[["sku","name","category","is_refurbished"]].head(10)
```

#Result:
The products were grouped into six clear business categories:
- Component (2370): internal parts such as RAM, storage, and hardware pieces
- Device (2188): main products like phones, tablets, laptops, and desktops
- Accessory (1983): cables, chargers, cases, stands, and similar items
- Other (1666): products that do not clearly fit into the main groups
- Network/Server (1112): NAS, servers, and networking equipment
- Peripheral (673): input/output devices like keyboards, mice, and monitors
